```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import het_arch
from statsmodels.stats.diagnostic import acorr_ljungbox
from pmdarima import auto_arima
from arch import arch_model

from sklearn.metrics import mean_squared_error, mean_absolute_error
import datetime
```

```python
# Parameters
TICKER = '^NSEI'  # Yahoo ticker for NIFTY 50
START = '2014-01-01'
END = '2016-12-31'
OUT_DIR = os.path.join('..', 'data')
os.makedirs(OUT_DIR, exist_ok=True)
print('Output dir:', OUT_DIR)
```

```python
# Download price data and save to CSV
df = yf.download(TICKER, start=START, end=END, progress=False)
if df.empty:
    raise RuntimeError('Data download returned empty DataFrame. Confirm ticker and network access.')
csv_path = os.path.join(OUT_DIR, 'nifty_2014_2016.csv')
df.to_csv(csv_path)
print('Saved:', csv_path)
df.head()
```

```python
df.reset_index(inplace=True)
df.columns = df.columns.droplevel(1)
```

```python
# Convert Date to datetime, drop rows with invalid dates, remove duplicates
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.dropna(subset=['Date'])
df = df.drop_duplicates(subset=['Date'])
df = df.sort_values('Date').reset_index(drop=True)
```

```python
df['LogPrice'] = np.log(df['Close'])
df['LogReturn'] = df['LogPrice'].diff()
df = df.dropna(subset=['LogReturn'])
```

```python
plt.figure(figsize=(10,4))
plt.plot(df['Close'])
plt.title("NIFTY 50 Close Price (2014-2016)")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

The plot shows the daily closing prices of the NIFTY 50 index from 2014 to 2016. The index exhibits significant fluctuations over the three-year period, reflecting the typical movement of a broad equity market. This behavior indicates the presence of market cycles, volatility, and non-stationarity in price levels, which is why returns (rather than prices) are typically used for time-series modelling.

```python
plt.figure(figsize=(10,4))
plt.plot(df['LogReturn'])
plt.title("NIFTY 50 Log Returns (2014-2016)")
plt.ylabel("Log Return")
plt.tight_layout()
plt.show()
```

The log-return series fluctuates around a mean close to zero, which is typical for financial return data. Unlike the price series, the returns do not display any clear long-term trend, confirming their stationarity. The plot also reveals occasional large spikes, which correspond to periods of heightened market volatility. Such volatility clustering—periods of high volatility followed by more high volatility—is a common feature in financial time series and justifies the use of ARCH/GARCH-type models later in the analysis.

```python
adf_res = adfuller(df['LogReturn'], autolag='AIC')
print("\nADF test statistic: {:.6f}".format(adf_res[0]))
print("ADF p-value: {:.6e}".format(adf_res[1]))
print("Used lags:", adf_res[2])
print("Number of observations:", adf_res[3])
print("Critical values:")
for k, v in adf_res[4].items():
    print(f"  {k}: {v:.6f}")
```

The ADF test strongly rejects the null hypothesis of a unit root (ADF statistic = –13.78 < –3.43 at 1% level, p-value ≈ 0). This confirms that the NIFTY 50 log-return series is stationary and suitable for ARIMA modeling.
ADF automatically selected 3 lags based on information criteria to remove autocorrelation in residuals.

```python
# ACF / PACF plots for returns
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plot_acf(df['LogReturn'], lags=40, ax=plt.gca(), title="ACF of Log Returns")
plt.subplot(1,2,2)
plot_pacf(df['LogReturn'], lags=40, ax=plt.gca(), method='ywm', title="PACF of Log Returns")
plt.tight_layout()
plt.show()
```

The ACF and PACF plots of the log returns show that almost all autocorrelations lie within the confidence bounds, except at lag 1. This indicates that the return series has very weak linear dependence, which is consistent with the efficient market hypothesis. The only spike at lag 1 suggests that a simple ARIMA(1,0,1) or ARMA(1,1) model is appropriate for capturing any remaining short-term dynamics in the return process. Overall, the plots confirm that the returns are largely uncorrelated but may still exhibit heteroskedastic behavior.

```python
# Train / test split (hold-out last h business days)
h = 60   # forecast horizon (you can change to 30 or 90)
train = df['LogReturn'].iloc[:-h].copy()
test  = df['LogReturn'].iloc[-h:].copy()

print("\nTrain length:", len(train), "Test length:", len(test))
```

```python
# ARIMA order selection on train (auto_arima)
print("\nRunning auto_arima (this may take a moment)...")
arima_auto = auto_arima(train, start_p=0, start_q=0, max_p=5, max_q=5,
                        seasonal=False, stepwise=True, suppress_warnings=True,
                        information_criterion='aic', trace=False)
order = arima_auto.order
print("Selected ARIMA order:", order)
```

```python
# Fit SARIMAX (statsmodels) to get residuals and diagnostics
sarimax_mod = sm.tsa.SARIMAX(train, order=order, enforce_stationarity=False, enforce_invertibility=False)
sarimax_res = sarimax_mod.fit(disp=False)
print("\nARIMA (SARIMAX) summary:")
print(sarimax_res.summary())
```

The **ARIMA (1,0,1) model fits the log-return series** reasonably well. Both the AR(1) and MA(1) coefficients are statistically significant (p-values < 0.05), indicating **short-term dependence in the returns**. The residual variance (sigma²) is small, reflecting the low average volatility of daily returns. However, the model diagnostics show non-normal residuals (significant Jarque–Bera test) and high kurtosis, which are common characteristics of financial returns. These observations suggest that, although the ARIMA model captures the mean dynamics effectively, it does not fully account for volatility clustering— **motivating the use of a GARCH model**.

```python
# 8) Residual diagnostics (ARIMA residuals)
resid = sarimax_res.resid.dropna()

plt.figure(figsize=(10,4))
plt.plot(resid)
plt.title("ARIMA Residuals")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,4))
plot_acf(resid, lags=40, ax=plt.gca(), title="ACF of ARIMA Residuals")
plt.tight_layout()
plt.show()
```

The residual plot appears to fluctuate randomly around zero without any visible pattern, which indicates that the ARIMA model has successfully captured the linear structure in the return series. Additionally, the ACF of residuals shows no significant autocorrelation at any lag, confirming that the residuals behave like white noise. This is further supported by the Ljung–Box test result, which fails to reject the null hypothesis of no autocorrelation. Therefore, the ARIMA model is adequate for modeling the conditional mean of the return series.

```python
# Ljung-Box on residuals
lb_res = acorr_ljungbox(resid, lags=[10], return_df=True)
print("\nLjung-Box test for residuals (lag 10):")
print(lb_res)
```

The Ljung–Box test has a very high p-value (≈ 0.96), which means we fail to reject the null hypothesis that residuals are uncorrelated. This confirms that the ARIMA model has successfully removed serial correlation from the return series. In other words, the mean dynamics are well modeled, and no significant autocorrelation remains. However, this test does not assess volatility patterns — only autocorrelation.

```python
# ARCH test on residuals (to justify GARCH)
arch_test_stat, arch_pvalue, _, _ = het_arch(resid, nlags=10)
print("\nARCH test (on residuals) p-value:", arch_pvalue)
if arch_pvalue < 0.05:
    print("-> Significant ARCH effects (justify GARCH).")
else:
    print("-> No strong ARCH evidence at 5% (but financial returns often show volatility clustering).")
```

The ARCH test produces a very small p-value (≈ 0.00057), indicating strong evidence of ARCH effects in the residuals. This means the variance of the residuals is not constant over time and exhibits volatility clustering. As a result, a GARCH model is justified to capture the time-varying volatility in the return series. This aligns with the stylized facts of financial markets, where periods of high volatility tend to cluster together.

```python
# Fit GARCH(1,1) on ARIMA residuals (scaled)
# Approach: fit GARCH on residuals (or directly on returns). We'll fit on residuals scaled by 100.
resid_scaled = (resid * 100).dropna()  # scale to percentages for numerical stability

garch = arch_model(resid_scaled, mean='Zero', vol='GARCH', p=1, q=1, dist='t')  # Zero mean since residuals
garch_res = garch.fit(disp='off')
print("\nGARCH(1,1) fit summary (on ARIMA residuals scaled by 100):")
print(garch_res.summary())
```

The GARCH(1,1) model fitted to the ARIMA residuals shows that volatility in the NIFTY-50 return series is highly persistent. Both α (0.0352) and β (0.9018) are statistically significant, and their sum (0.937) is close to 1, indicating strong volatility clustering and slow decay of shocks—common properties of financial returns. The degrees of freedom parameter (ν = 6.62) from the Student-t distribution suggests heavy tails, meaning extreme return movements occur more frequently than under the normal distribution. This is consistent with the QQ-plot and the statistically significant ARCH test, which confirmed the presence of conditional heteroskedasticity. Overall, the GARCH(1,1) model appropriately captures the time-varying volatility structure of the NIFTY 50 returns and provides a realistic estimate of risk and volatility forecasts.

```python
# Conditional volatility (in original return units)
cond_vol_scaled = garch_res.conditional_volatility  # percent units (since we scaled)
cond_vol = cond_vol_scaled / 100.0  # back to returns units (daily std)
cond_vol.index = resid.index

plt.figure(figsize=(10,4))
plt.plot(cond_vol)
plt.title("Conditional Volatility (GARCH(1,1)) - daily std dev")
plt.ylabel("Volatility (daily std dev)")
plt.tight_layout()
plt.show()
```

The conditional volatility plot obtained from the GARCH(1,1) model shows how the daily volatility of NIFTY 50 returns evolves over time. The volatility is clearly time-varying, with noticeable spikes during periods of market stress (for example, mid-2015 and early 2016). These spikes represent times when returns were more uncertain and markets were more turbulent. The presence of such volatility clustering confirms that the GARCH model is appropriate for this dataset. Overall, the model captures the characteristic behavior of financial markets where volatility rises and falls in clusters rather than remaining constant.

```python
# Forecast volatility with GARCH
# We'll forecast volatility for horizon 'h' with the GARCH fit (on residuals)
garch_fore = garch_res.forecast(horizon=h, reindex=False)  # returns object with .variance
# Extract h-step ahead variance forecast from last available (use .variance)
# garch_fore.variance is ndarray-like: shape (1, h) for the last roll
var_fore = garch_fore.variance.values[-1, :]  # variance in percent^2 units
vol_fore = np.sqrt(var_fore) / 100.0         # convert back to return units (daily std dev)
# Create series indexed by test dates
vol_fore_series = pd.Series(vol_fore, index=test.index)

plt.figure(figsize=(10,4))
plt.plot(vol_fore_series, label='GARCH 1-step...h-step volatility forecast (daily std dev)')
plt.title("Volatility Forecast (GARCH)")
plt.ylabel("Forecasted daily volatility")
plt.legend()
plt.tight_layout()
plt.show()
```

The volatility forecast from the GARCH(1,1) model shows a smooth, gradually increasing pattern over the forecast horizon. This is typical of GARCH forecasts, where future volatility tends to revert toward a long-run average value. Because recent periods in the data had moderately high volatility, the model anticipates slightly elevated levels in the near future. The absence of sudden jumps in the forecast reflects the fact that no new shock information enters during forecasting; the model propagates volatility forward based solely on its internal dynamics.

```python
# 12) ARIMA residual diagnostics after GARCH (optional)
# If desired, examine standardized residuals of GARCH
std_resid = garch_res.std_resid  # standardized residuals (should be approximately iid t-dist)
plt.figure(figsize=(10,4))
plt.plot(std_resid)
plt.title("GARCH Standardized Residuals")
plt.tight_layout()
plt.show()
```

The standardized residuals represent the GARCH-filtered return series, where the effects of changing volatility have been removed. These residuals appear to fluctuate randomly around zero, with no obvious pattern or clustering. This indicates that the GARCH model has effectively captured the heteroskedasticity in the data. The remaining noise behaves more like an independent and identically distributed (i.i.d.) series, which is a desirable property. Occasional large spikes still appear, reflecting the use of a heavy-tailed distribution (Student-t), which appropriately models extreme financial movements.

```python
# QQ-plot of standardized residuals
sm.qqplot(std_resid, line='s')
plt.title("QQ plot - GARCH standardized residuals")
plt.tight_layout()
plt.show()
```

The QQ-plot of GARCH standardized residuals shows moderate alignment with the normal distribution in the central quantiles, but significant deviations in the tails. This indicates that the residuals exhibit heavy-tailed behavior, which is typical for financial return series. Therefore, the residuals are not perfectly normally distributed, and a heavy-tailed distribution such as Student-t is more appropriate.

```python
#clipping data into train and test for walk-forward one step forecasting
train = df[:'2016-07-31'].copy()
test  = df['2016-08-01':].copy()
```

```python
# Now walk-forward ARIMA/SARIMAX on log-price (one-step ahead)
# choosing ARIMA order on training log-price (auto_arima)

stepwise = True
print("Selecting ARIMA order on log-price")
model_auto = auto_arima(train['LogPrice'], seasonal=False, stepwise=stepwise,
                        max_p=4, max_q=4, d=None, trace=False, error_action='ignore', suppress_warnings=True)
order = model_auto.order
print("Selected order (p,d,q):", order)
```

```python
# Walk-forward one-step forecasting just like Teacher Forcing in Deep Learning:
history = train['LogPrice'].copy()
pred_logprice = []
for idx, row in test.iterrows():
    # fit SARIMAX on history (small models are fast)
    mod = sm.tsa.SARIMAX(history, order=order, enforce_stationarity=False, enforce_invertibility=False)
    res = mod.fit(disp=False)
    # one-step forecast (log-price)
    f = res.get_forecast(steps=1)
    pred = f.predicted_mean.iloc[0]
    pred_logprice.append(pred)
    # now append the actual observed logprice to history for next iteration (rolling update)
    history = pd.concat([history, pd.Series([row['LogPrice']], index=[idx])])


# attach predictions to test
test = test.copy()
test['pred_logprice'] = pred_logprice
test['pred_price'] = np.exp(test['pred_logprice'])
test['true_price'] = np.exp(test['LogPrice'])

# metrics
rmse_price = np.sqrt(mean_squared_error(test['true_price'], test['pred_price']))
mae_price  = mean_absolute_error(test['true_price'], test['pred_price'])
print(f"RMSE (price): {rmse_price:.2f}, MAE (price): {mae_price:.2f}")
```
